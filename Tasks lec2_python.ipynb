2.  Compare the values of my_age and your_age using **`if-else`**. Who is older (me or you)? Use **`input(“Enter your age: ”)`** to get the age as input. You can use a nested condition to print **`'year'`** for 1 year difference in age, **`'years'`** for bigger differences, and a custom text if **`my_age = your_age`**.

​
Output:
​
```py

```

````python
my_age  = 25
your_age = int(input("Enter your age: "))

if your_age > my_age:
    if your_age - my_age == 1:
        print(f'You are {your_age - my_age} year older than me.')
    else:
        print(f'You are {your_age - my_age} years older than me.')
elif your_age < my_age:
    if my_age - your_age == 1:
        print(f'I am {my_age - your_age} year older than you.')
    else:
        print(f'I am {my_age - your_age} years older than you.')
else:
    print("We are the same age.")
````

3. Check if the season is Autumn, Winter, Spring or Summer.
    - If the user input is:
        - September, October or November, the season is Autumn.
        - December, January or February, the season is Winter.
        - March, April or May, the season is Spring
        - June, July or August, the season is Summer


````python
month = input("Enter the month: ").lower()
if month in ['september', 'october', 'november']:
    print("The season is Autumn.")
elif month in ['december', 'january', 'february']:
    print("The season is Winter.")
elif month in ['march', 'april', 'may']:
    print("The season is Spring.")
elif month in ['june', 'july', 'august']:
    print("The season is Summer.")
else:
    print("Invalid month.")
````

# Part 2: Dictionary & Conditions Practice


You are given a dictionary called `person` that contains information about an individual.  
Your task is to perform the following operations step-by-step:

##  The Dictionary:

```python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
```

## Tasks

### 1. Check for `skills` key
- If the `skills` key exists in the dictionary:
  - Print the **middle skill** from the list of skills.

### 2. Check for `'Python'` in skills
- Still inside the same condition:
  - Check if `'Python'` is in the person's skills list.
  - Print `True` if found, otherwise `False`.

### 3. Based on the combination of skills:
  - If the person has **only** `['Python', 'Matlab']` → print: `He knows machine learning`
  - If the person has both `'Python'` and `'R'` → print: `He knows statistics`
  - If the person has both `'C'` and `'C++'` → print: `He knows software development`
  - Else → print: `unknown title`


### 4. Print location and marital status
- If the person is married **and** lives in **Finland**, print:
```py
    Milaan Parmar lives in Finland. He is married.
```



````python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}

try:
    if 'skills' in person.keys():
        print(f"Middle Skill: {person['skills'][len(person['skills'])//2]}")  # Accessing the middle skill
    else:
        print("Skills key is not present in the person dictionary.")
except KeyError as e:
    print(f"Error: Key not found - {e}")
````

````python
if person.get('skills'):
    if 'Python' in person['skills']:
        print("True")
    else:
        print("False.")
````

````python
if ['python', 'Matlab'] == person.get('skills'):
    print("He knows machine learning.")
elif  ['python', 'R'] in person.get('skills'):
    print("He knows Statistics.")
elif ['c', 'C++'] in person.get('skills'):
    print("He knows software development.")
else:
    print("Unknown title.")
````

````python
if person['country'].lower() == 'finland' and person['is_marred'] == True:
    print(f"{person['first_name']} {person['last_name']} lives in {person['country']}. He is married.")
````
